```python
import pandas as pd 
```

```python
data=pd.read_csv("Netflix_Userbase_Updated.csv")
data.head()
```

```python
print(data.isnull().sum())
```

```python
data.shape
```

```python
data["Country"].fillna(data["Country"].mode()[0],inplace=True)
```

```python
print(data.isnull().sum())
```

```python
import matplotlib.pyplot as plt 
plt.hist(data["Age"],bins=8,edgecolor="black")
plt.show()
```

```python
print("Total:",data.groupby("Country")["Monthly Revenue"].sum())
```

```python
plt.scatter(data["Age"],data["Monthly Revenue"])
plt.show()
```

```python
import seaborn as sns
sns.countplot(data=data,color="Red",x="Country")
plt.xticks(rotation=25, ha="right")
plt.show()
```

```python
sns.countplot(data=data,color="Blue",x="Subscription Type")
plt.xticks(rotation=25, ha="right")
plt.show()
```

```python
plt.figure(figsize=(30,10))
sns.lineplot(data=data,color="green",x="Country",y="Subscription Type")
plt.show()
```

```python
plt.figure(figsize=(10,15))
sns.lineplot(data=data,color="Orange",x="Age",y="Subscription Type")
plt.show()
```

```python
plt.figure(figsize=(15,5))
sns.countplot(data=data,x="Gender")
plt.show()
```
